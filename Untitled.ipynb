## Map Purchase Frequency to Numbers

```python
# Map text frequency into numeric score
freq_map = {
    'Multiple times a week'  : 5,
    'Once a week'            : 4,
    'Few times a month'      : 3,
    'Once a month'           : 2,
    'Less than once a month' : 1
}

df['Freq_Score'] = df['Purchase_Frequency'].map(freq_map)

print(df['Freq_Score'].value_counts())
print("\nAny nulls? :", df['Freq_Score'].isnull().sum())
```

## Map Cart Completion into Numbers

```python
cart_map = {
    'Always' :5,
    'Often' : 4,
    'Sometimes' : 3,
    'Rarely' : 2,
    'Never' : 1
}
df['Cart_Completion_Score'] = df['Cart_Completion_Frequency'].map(cart_map)
print(df['Cart_Completion_Score'].value_counts())
```

## Customer Segmentation based on shopping

```python
def segment_customer(row):
    if row ['Freq_Score'] >= 4 and row['Shopping_Satisfaction'] >= 4:
        return 'Frequent Buyer'
    elif row['Shopping_Satisfaction'] <= 2 or row['Cart_Completion_Score'] <=2:
        return 'At Risk Customer'
    else:
        return 'Occasional Buyer'
df['Segment'] = df.apply(segment_customer,axis=1)
print(df['Segment'].value_counts())
```

## K-Means Clustring

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
cluster_columns = ['Freq_Score','Shopping_Satisfaction','Customer_Reviews_Importance','Rating_Accuracy','Cart_Completion_Score']
X = df[cluster_columns].dropna()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

inertia = []
for k in range(2,9):
    km = KMeans(n_clusters = k, random_state = 42, n_init = 10)
    km.fit(X_scaled)
    inertia.append(km.inertia_)
plt.figure(figsize = (7,4))
plt.plot(range(2,9), inertia, marker = 'o', color = 'mediumvioletred')
plt.title('Finding Optimal number of Clusters')
plt.xlabel('Number of Clusters k')
plt.ylabel('Inertia')
plt.tight_layout()
plt.show()
```

```python
km = KMeans(n_clusters=3, random_state=42, n_init=10)
df.loc[X.index, 'KMeans_Cluster'] = km.fit_predict(X_scaled)
df['KMeans_Cluster'] = df['KMeans_Cluster'].astype('Int64')

print("K-Means Cluster Counts:")
print(df['KMeans_Cluster'].value_counts())
```

```python
# plot between Freq vs Satisfaction
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Freq_Score', y='Shopping_Satisfaction',
                hue='KMeans_Cluster', palette='Set1', s=60, alpha=0.7)
plt.title('K-Means Clusters: Purchase Frequency vs Satisfaction')
plt.xlabel('Purchase Frequency Score')
plt.ylabel('Shopping Satisfaction')
plt.legend(title='Cluster')
plt.tight_layout()
plt.show()

# Average metrics per cluster
cluster_stats = df.groupby('KMeans_Cluster')[cluster_columns].mean().round(2)
print("Cluster Profiles")
print(cluster_stats)
```

```python
print("Rule-Based Segments vs K-Means Clusters")
print(pd.crosstab(df['Segment'], df['KMeans_Cluster']))
```

```python
print("""
Comparison Insights:
- Occasional Shoppers show strongest agreement with K-Means Cluster 0
- Frequent Buyers are captured mostly in Clusters 1 and 2
- At-Risk Customers are distributed across all clusters, suggesting
  K-Means captures more nuanced behavioral patterns beyond just
  satisfaction and cart completion scores
- Overall, both methods complement each other — rule-based gives
  business-interpretable labels while K-Means reveals hidden patterns
""")
```

## Correlation Heatmap

```python
# Encode categorical columns for correlation
encode_map = {
    'Recommendation_Helpfulness' : {'Yes': 3, 'Sometimes': 2, 'No': 1},
    'Review_Helpfulness'         : {'Yes': 3, 'Sometimes': 2, 'No': 1},
    'Review_Left'                : {'Yes': 1, 'No': 0},
    'Review_Reliability'         : {'Heavily': 5, 'Moderately': 4,
                                    'Occasionally': 3, 'Rarely': 2, 'Never': 1},
    'Personalized_Recommendation_Frequency': {'Yes': 3, 'Sometimes': 2, 'No': 1}
}

df_corr = df.copy()
for col, mapping in encode_map.items():
    df_corr[col] = df_corr[col].map(mapping)

corr_cols = ['Shopping_Satisfaction', 'Rating_Accuracy',
             'Customer_Reviews_Importance', 'Recommended_Frequency_Score',
             'Recommendation_Helpfulness', 'Review_Helpfulness',
             'Review_Reliability', 'Review_Left',
             'Personalized_Recommendation_Frequency']

plt.figure(figsize=(10, 7))
sns.heatmap(df_corr[corr_cols].corr(), 
            annot=True, fmt='.2f', 
            cmap='coolwarm', 
            center=0)
plt.title('Correlation: Review & Recommendation Metrics')
plt.tight_layout()
plt.show()
```
