```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from collections import Counter
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('Amazon_cleaned.csv')

# Global style settings for all charts
sns.set_theme(style='whitegrid')
plt.rcParams['font.family']  = 'DejaVu Sans'
plt.rcParams['axes.titlesize'] = 13
plt.rcParams['axes.titleweight'] = 'bold'

print("Ready. Shape:", df.shape)
```

```python
# Split semicolon-separated categories
all_cats = []
for entry in df['Purchase_Categories'].dropna():
    cats = [c.strip() for c in str(entry).split(';')]
    all_cats.extend(cats)

cat_df = pd.DataFrame(
    Counter(all_cats).most_common(), 
    columns=['Category', 'Count'])

plt.figure(figsize=(9, 5))
bars = plt.barh(cat_df['Category'], cat_df['Count'],
                color=sns.color_palette('viridis', len(cat_df)))
plt.xlabel('Number of Customers')
plt.title('Most Popular Purchase Categories on Amazon')
plt.gca().invert_yaxis()  # Highest on top

# Add count labels on bars
for bar in bars:
    plt.text(bar.get_width() + 5, bar.get_y() + bar.get_height()/2,
             str(int(bar.get_width())),
             va='center', fontsize=9)

plt.tight_layout()
plt.savefig('chart1_purchase_categories.png')
plt.show()
```

```python
browse_order  = ['Multiple times a day', 'Few times a week',
                 'Few times a month', 'Rarely']
browse_counts = df['Browsing_Frequency'].value_counts().reindex(browse_order)
colors        = ['#2196F3', '#4CAF50', '#FF9800', '#F44336']
explode       = (0.05, 0.05, 0.05, 0.05)

plt.figure(figsize=(7, 6))
wedges, texts, autotexts = plt.pie(
    browse_counts,
    labels=browse_order,
    autopct='%1.1f%%',
    colors=colors,
    explode=explode,
    startangle=140,
    textprops={'fontsize': 11})

for autotext in autotexts:
    autotext.set_fontweight('bold')

plt.title('Browsing Frequency Distribution', pad=20)
plt.tight_layout()
plt.savefig('chart2_browsing_frequency.png')
plt.show()
```

```python
sat_counts = df['Shopping_Satisfaction'].value_counts().sort_index()
colors_sat  = ['#F44336','#FF9800','#FFC107','#8BC34A','#4CAF50']
labels_sat  = ['1 - Very Low','2 - Low','3 - Moderate',
               '4 - High','5 - Very High']

plt.figure(figsize=(8, 5))
bars = plt.bar(labels_sat, sat_counts.values,
               color=colors_sat, edgecolor='black', width=0.6)

# Add count + percentage labels on top of bars
total = len(df)
for bar, count in zip(bars, sat_counts.values):
    pct = count / total * 100
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 3,
             f'{count}\n({pct:.1f}%)',
             ha='center', va='bottom', fontsize=9)

plt.title('Shopping Satisfaction Level Distribution')
plt.xlabel('Satisfaction Level')
plt.ylabel('Number of Customers')
plt.xticks(rotation=15, ha='right')
plt.tight_layout()
plt.savefig('chart3_satisfaction_levels.png')
plt.show()
```

```python
help_order = ['Yes', 'Sometimes', 'No']

# Count of each satisfaction score within each helpfulness group
rec_sat = df.groupby(['Recommendation_Helpfulness', 
                       'Shopping_Satisfaction']).size().unstack(fill_value=0)
rec_sat = rec_sat.reindex(help_order)

plt.figure(figsize=(9, 5))
rec_sat.plot(kind='bar',
             colormap='RdYlGn',
             edgecolor='black',
             ax=plt.gca())

plt.title('Recommendation Helpfulness vs Shopping Satisfaction')
plt.xlabel('Recommendation Helpful?')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.legend(title='Satisfaction Score', bbox_to_anchor=(1.01, 1))
plt.tight_layout()
plt.savefig('chart4_rec_vs_satisfaction.png')
plt.show()
```

```python
# Encode categoricals for correlation
encode_map = {
    'Recommendation_Helpfulness'            : {'Yes':3,'Sometimes':2,'No':1},
    'Review_Helpfulness'                    : {'Yes':3,'Sometimes':2,'No':1},
    'Review_Left'                           : {'Yes':1,'No':0},
    'Review_Reliability'                    : {'Heavily':5,'Moderately':4,
                                               'Occasionally':3,'Rarely':2,'Never':1},
    'Personalized_Recommendation_Frequency' : {'Yes':3,'Sometimes':2,'No':1}
}

df_corr = df.copy()
for col, mapping in encode_map.items():
    df_corr[col] = df_corr[col].map(mapping)

corr_cols = ['Shopping_Satisfaction','Rating_Accuracy',
             'Customer_Reviews_Importance','Recommended_Frequency_Score',
             'Recommendation_Helpfulness','Review_Helpfulness',
             'Review_Reliability','Review_Left',
             'Personalized_Recommendation_Frequency']

corr_matrix = df_corr[corr_cols].corr()

plt.figure(figsize=(10, 7))
mask = np.zeros_like(corr_matrix, dtype=bool)
mask[np.triu_indices_from(mask)] = True  # Hide upper triangle

sns.heatmap(corr_matrix,
            annot=True, fmt='.2f',
            cmap='coolwarm', center=0,
            mask=mask,
            linewidths=0.5,
            cbar_kws={'shrink': 0.8})

plt.title('Correlation Heatmap — Review & Recommendation Metrics')
plt.tight_layout()
plt.savefig('chart5_correlation_heatmap.png')
plt.show()
```

```python
fig = plt.figure(figsize=(20, 24))
fig.suptitle('Amazon Customer Behaviour',
             fontsize=20, fontweight='bold', y=0.98)

gs = gridspec.GridSpec(3, 2, figure=fig)

#Chart 1: Purchase Categories
ax1 = fig.add_subplot(gs[0, 0])
ax1.barh(cat_df['Category'], cat_df['Count'],
         color=sns.color_palette('viridis', len(cat_df)))
ax1.invert_yaxis()
ax1.set_title('Most Popular Purchase Categories')
ax1.set_xlabel('Count')

# Chart 2: Browsing Frequency
ax2 = fig.add_subplot(gs[0, 1])
ax2.pie(browse_counts,
        labels=browse_order,
        autopct='%1.1f%%',
        colors=colors,
        startangle=140,
        textprops={'fontsize': 9})
ax2.set_title('Browsing Frequency Distribution')

# Chart 3: Satisfaction Levels
ax3 = fig.add_subplot(gs[1, 0])
bars3 = ax3.bar(labels_sat, sat_counts.values,
                color=colors_sat, edgecolor='black')
ax3.set_title('Shopping Satisfaction Distribution')
ax3.set_ylabel('Count')
ax3.set_xticklabels(labels_sat, rotation=15, ha='right', fontsize=8)

# Chart 4: Rec Helpfulness vs Satisfaction
ax4 = fig.add_subplot(gs[1, 1])
rec_sat_means = df.groupby('Recommendation_Helpfulness')['Shopping_Satisfaction']\
                  .mean().reindex(help_order)
bars4 = ax4.bar(help_order, rec_sat_means,
                color=['mediumseagreen','gold','tomato'],
                edgecolor='black')
ax4.set_title('Avg Satisfaction by Recommendation Helpfulness')
ax4.set_ylabel('Avg Satisfaction Score')
ax4.set_ylim(0, 6)

# Chart 5: Correlation Heatmap
ax5 = fig.add_subplot(gs[2, :])  # Full bottom row
sns.heatmap(corr_matrix,
            annot=True, fmt='.2f',
            cmap='coolwarm', center=0,
            linewidths=0.5,
            ax=ax5,
            cbar_kws={'shrink': 0.6})
ax5.set_title('Correlation Heatmap — Review & Recommendation Metrics')

plt.savefig('amazon_dashboard.png')
plt.show()
```
